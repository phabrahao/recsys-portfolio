# Node 1 — EDA: MovieLens 32M

Purpose: not general exploration — three charts that directly inform design decisions in later nodes:

1. **Ratings volume over time** → confirms the timestamp range/density needed for the point-in-time leakage test.
2. **Ratings per movie (long tail)** → the cold-start distribution that node 3's ranking model needs to handle.
3. **Ratings per user** → same idea, user-side sparsity — relevant for how much history a "new session" behavioral feature can lean on.

Schema (confirmed from the raw CSVs):
- `ratings.csv`: userId, movieId, rating, timestamp (unix seconds)
- `movies.csv`: movieId, title, genres (pipe-separated)
- `links.csv`: movieId, imdbId, tmdbId
- `tags.csv`: userId, movieId, tag, timestamp

```python
import polars as pl
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from pathlib import Path
from datetime import datetime

DATA_DIR = Path("../data/raw/ml-32m")  # adjust if your notebook lives elsewhere
CHARTS_DIR = Path("charts")
CHARTS_DIR.mkdir(exist_ok=True)

plt.rcParams["figure.dpi"] = 150
```

```python
# Lazy scan — don't materialize 32M rows until we actually need to.
ratings_lazy = pl.scan_csv(DATA_DIR / "ratings.csv")

print(ratings_lazy.collect_schema())
n_rows = ratings_lazy.select(pl.len()).collect().item()
print(f"ratings.csv: {n_rows:,} rows")
```

## 1. Ratings volume over time
Confirms the timestamp span and checks for gaps/bursts before we build the synthetic event stream on top of it.

```python
daily_counts = (
    ratings_lazy
    .with_columns(
        pl.from_epoch("timestamp", time_unit="s").dt.date().alias("date")
    )
    .group_by("date")
    .agg(pl.len().alias("n_ratings"))
    .sort("date")
    .collect()
)

print(f"Date range: {daily_counts['date'].min()} → {daily_counts['date'].max()}")

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(daily_counts["date"], daily_counts["n_ratings"], linewidth=0.6)
ax.set_title("Ratings volume per day")
ax.set_xlabel("Date")
ax.set_ylabel("Number of ratings")
ax.xaxis.set_major_locator(mdates.YearLocator(2))
fig.tight_layout()
fig.savefig(CHARTS_DIR / "ratings_volume_over_time.png", bbox_inches="tight")
plt.show()
```

## 2. Ratings per movie (cold-start / long tail)
Log-scale x-axis — this is the chart that justifies content-based cold-start scoring in node 3.

```python
ratings_per_movie = (
    ratings_lazy
    .group_by("movieId")
    .agg(pl.len().alias("n_ratings"))
    .collect()
)

median_r = ratings_per_movie["n_ratings"].median()
pct_under_10 = (ratings_per_movie["n_ratings"] < 10).mean() * 100
print(f"Median ratings/movie: {median_r:.0f}")
print(f"Movies with < 10 ratings: {pct_under_10:.1f}%  <- your cold-start population")

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(ratings_per_movie["n_ratings"], bins=50, log=True)
ax.set_xscale("log")
ax.set_title("Ratings per movie (long tail)")
ax.set_xlabel("Number of ratings (log scale)")
ax.set_ylabel("Number of movies (log scale)")
fig.tight_layout()
fig.savefig(CHARTS_DIR / "ratings_per_movie.png", bbox_inches="tight")
plt.show()
```

## 3. Ratings per user
Same idea, user side — informs how much interaction history a "new session" feature can realistically lean on.

```python
ratings_per_user = (
    ratings_lazy
    .group_by("userId")
    .agg(pl.len().alias("n_ratings"))
    .collect()
)

median_u = ratings_per_user["n_ratings"].median()
pct_under_10_u = (ratings_per_user["n_ratings"] < 10).mean() * 100
print(f"Median ratings/user: {median_u:.0f}")
print(f"Users with < 10 ratings: {pct_under_10_u:.1f}%")

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(ratings_per_user["n_ratings"], bins=50, log=True)
ax.set_xscale("log")
ax.set_title("Ratings per user (long tail)")
ax.set_xlabel("Number of ratings (log scale)")
ax.set_ylabel("Number of users (log scale)")
fig.tight_layout()
fig.savefig(CHARTS_DIR / "ratings_per_user.png", bbox_inches="tight")
plt.show()
```
